```python
#Minimax
import math

def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):
    if curDepth == targetDepth:
        return scores[nodeIndex]

    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2, False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, False, scores, targetDepth)
        )
    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2, True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, True, scores, targetDepth)
        )

scores = [3, 5, 2, 9, 0, 7]
treeDepth = math.ceil(math.log2(len(scores)))

scores += [0] * (2**treeDepth - len(scores))

print("Scores:", scores)
print("The optimal value is:", minimax(0, 0, True, scores, treeDepth))
```

```python
#Minimax
import math

count = 0

def minimax(curDepth, nodeIndex, maxTurn, scores, targetDepth):
    global count

    if curDepth == targetDepth:
        count += 1
        return scores[nodeIndex]

    if maxTurn:
        return max(
            minimax(curDepth + 1, nodeIndex * 2, False, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, False, scores, targetDepth)
        )
    else:
        return min(
            minimax(curDepth + 1, nodeIndex * 2, True, scores, targetDepth),
            minimax(curDepth + 1, nodeIndex * 2 + 1, True, scores, targetDepth)
        )

scores = [3, 5, 2, 9, 0, 7]
treeDepth = math.ceil(math.log2(len(scores)))
scores += [0] * (2**treeDepth - len(scores))

result = minimax(0, 0, True, scores, treeDepth)

print("Optimal value:", result)
print("Evaluated leaves:", count)
```

```python
#alphabeta
import math

count_ab = 0

def alphabeta(depth, index, maxTurn, scores, target, alpha, beta):
    global count_ab

    if depth == target:
        count_ab += 1
        return scores[index]

    if maxTurn:
        value = float('-inf')

        for i in range(2):
            value = max(
                value,
                alphabeta(depth + 1, index * 2 + i, False,
                          scores, target, alpha, beta)
            )
            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:
        value = float('inf')

        for i in range(2):
            value = min(
                value,
                alphabeta(depth + 1, index * 2 + i, True,
                          scores, target, alpha, beta)
            )
            beta = min(beta, value)

            if alpha >= beta:
                break

        return value


scores = [3, 5, 2, 9, 3, 5, 2, 9]
depth = int(math.log2(len(scores)))

result_ab = alphabeta(
    0, 0, True, scores, depth,
    float('-inf'), float('inf')
)

print("Alpha-Beta optimal value:", result_ab)
print("Evaluated leaves:", count_ab)
```

```python
#Comparison
def minimax_count(depth, index, maxTurn, scores, target, count):
    if depth == target:
        count[0] += 1
        return scores[index]

    if maxTurn:
        return max(
            minimax_count(depth+1, index*2, False, scores, target, count),
            minimax_count(depth+1, index*2+1, False, scores, target, count)
        )
    else:
        return min(
            minimax_count(depth+1, index*2, True, scores, target, count),
            minimax_count(depth+1, index*2+1, True, scores, target, count)
        )


def alphabeta_count(depth, index, maxTurn, scores, target,
                    alpha, beta, count):
    if depth == target:
        count[0] += 1
        return scores[index]

    if maxTurn:
        value = float('-inf')
        for i in range(2):
            value = max(value, alphabeta_count(
                depth+1, index*2+i, False, scores, target,
                alpha, beta, count
            ))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    else:
        value = float('inf')
        for i in range(2):
            value = min(value, alphabeta_count(
                depth+1, index*2+i, True, scores, target,
                alpha, beta, count
            ))
            beta = min(beta, value)
            if alpha >= beta:
                break
        return value


orders = [
    [3, 5, 2, 9, 3, 5, 2, 9],
    [2, 9, 3, 5, 2, 9, 3, 5]
]

for scores in orders:
    depth = 3
    mc = [0]
    ac = [0]

    m = minimax_count(0, 0, True, scores, depth, mc)
    a = alphabeta_count(
        0, 0, True, scores, depth,
        float('-inf'), float('inf'), ac
    )

    print("\nScores:", scores)
    print("Minimax value:", m, "| Leaves:", mc[0])
    print("Alpha-Beta value:", a, "| Leaves:", ac[0])
```
